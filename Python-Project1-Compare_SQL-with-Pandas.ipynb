#### Memory optimisation for first dataset
- Reduce the memory usage by storing some columns as a category type 
- The category type is suitable for columns with a limited number of unique values

### Filtering with multiple conditions (AND Logic)

```python
# show all people with diabetes and have a high sugar consumption
has_diabetes = heart_dis["Diabetes"] == "Yes"
heart_dis[has_diabetes]

has_high_sugar_consumption = heart_dis["Sugar Consumption"] == "High"
heart_dis[has_high_sugar_consumption]

heart_dis[has_diabetes & has_high_sugar_consumption]
```

### Filtering data with either/or conditions (OR Logic)

```python
# show all people that either smoke or have a high alcohol consumption

smokes  = heart_dis["Smoking"] == "Yes"

high_alcohol = heart_dis["Alcohol Consumption"] == "High"

heart_dis[smokes | high_alcohol]
```

### Data Manipulation
The second heart attack risk dataset has a patient id to uniquely identify each record but the first heart disease dataset doesn't. In order to illustrate the join examples, a subset of the first heart disease dataset will be modified to include a synthetic patient id.
Two additional datasets will be used and these are subsets of the heart attack risk dataset.
- female_data
- male_data
- Both dataset have the following fields: patient id, Age and Sex 


##### Step 1
Extract a subset of heart disease data and add the patient ids from the female_data

##### 1.1 Clean the heart disease age column

```python
heart_dis["Age"].sort_values() # sorting the age column shows that there are missing values

# remove the missing values before converting it to Integer to match the heart attack data
heart_dis_subset = heart_dis.dropna(subset=["Age"]) # drops the rows for the specific column where the entries are missing
heart_dis_subset.loc[:,"Age"] = heart_dis_subset["Age"].astype("int") # Update the datatype of the Age from Float ==> integer
heart_dis_subset.head(10)
```

##### 1.2 filter out a subset of the data based on the following criteria and modify the data.
- female (sorted in age ascending order)
- get a subset

```python
# filter out the female in the data and sort by age
# get the top 33 in the list and add the synthetic patient id to the data

is_female = heart_dis_subset["Gender"] == "Female"
eighteen_years_female = heart_dis_subset[is_female].sort_values(["Age"]).head(33)
```

##### 1.3 Add the synthetic patient id to the subset

```python
# combine the synthetic data with the patient id to the created subset

# read in the CSV
# save only the patient id
# concatenate the patient id to the rest of the dataset

list_with_patient_id_1 = pd.read_csv("../notebooks/project-dataset/female_data.csv")
list_with_patient_id_1 = list_with_patient_id_1["Patient ID"]

list_with_patient_id_1 = pd.concat([list_with_patient_id_1.reset_index(drop=True), eighteen_years_female.reset_index(drop=True)], axis=1)
```

##### Step 2
Extract a subset of heart disease data and add the patient ids from the male_data

##### 2.1 filter out a subset of the data based on the following criteria and modify the data.
- male (sorted in age ascending order)
- get a subset

```python
# filter out the male in the data and sort by age
# get the top 33 in the list and add the synthetic patient id to the data

is_male = heart_dis_subset["Gender"] == "Male"
eighteen_years_male = heart_dis_subset[is_male].sort_values(["Age"]).head(33)
```

##### 2.2 Add the synthetic patient id to the subset

```python
# combine the synthetic data with the patient id to the created subset

# read in the CSV
# save only the patient id
# concatenate the patient id to the rest of the dataset

list_with_patient_id_2 = pd.read_csv("../notebooks/project-dataset/male_data.csv")
list_with_patient_id_2 = list_with_patient_id_2["Patient ID"]

list_with_patient_id_2 = pd.concat([list_with_patient_id_2.reset_index(drop=True), eighteen_years_male.reset_index(drop=True)], axis=1)
```

##### 2.3 Finally combine the two lists to make the new heart disease subset to illustrate joins

```python
heart_disease_with_patient_id = pd.concat([list_with_patient_id_1, list_with_patient_id_2], ignore_index=True).sort_index()
heart_disease_with_patient_id
```

### Comparison 6

#### Joins

This section illustrates different joins possible in python

For the purpose of these examples, the ***left table*** is the ***heart attack risk data*** and the ***right table*** is the ***heart disease data***

| Join Type     | Description
|-----------    |-----------------------------------------------------------------------------------------------------------------
| Inner Join    | Show patients that are in both datasets
| Left Join     | Show all the patients that are in the heart attack risk, and only those that match from the heart disease data
| Right Join    | Show all the patients from the heart disease data and only those that match from the heart attack risk.
| Outer Join    | Show all the patients from both heart disease data and heart attack risk data and match them where possible.



**Inner Join** <br>
Show patients that are in both datasets
- Display all the columns
- Display a subset of the columns - Age, Sex, Cholesterol level, Blood Pressure, Heart Rate, Family History

```python
# Show patients that are in both datasets and display all the columns
patients_in_both_datasets = heart_att.merge(heart_disease_with_patient_id, how="inner", on="Patient ID", suffixes=[" - HeartAttackData", " - HeartDiseaseData"])
patients_in_both_datasets #36 rows
```

```python
# Show patients that are in both datasets and display a subset of the columns:  Age, Gender, Cholesterol level, Blood Pressure, Family History
patients_in_both_datasets[["Patient ID", "Age - HeartAttackData", "Gender", "Cholesterol", "Cholesterol Level", "Blood Pressure - HeartAttackData", "Blood Pressure - HeartDiseaseData", "Family History", "Family Heart Disease"]]
```

**Left Join**<br>

Show all the patients that are in the heart attack risk (left table), and only those that match from the heart disease data (right table)
- Display all the columns
- Display a subset of the columns - Age, Sex, Cholesterol level, Blood Pressure, Heart Rate, Family History

```python
# Show all the patients that are in the heart attack risk (left table), and only those that match from the heart disease data (right table) and display all the columns
left_join_on_datasets = heart_att.merge(heart_disease_with_patient_id, how="left", on="Patient ID", suffixes=[" - HeartAttackData", " - HeartDiseaseData"])
left_join_on_datasets #8,763 rows
```

```python
# Show all the patients that are in the heart attack risk (left table), and only those that match from the heart disease data (right table) and display a subset of the columns:  
# Age, Gender, Cholesterol level, Blood Pressure, Family History
left_join_on_datasets[["Patient ID", "Age - HeartAttackData", "Gender", "Cholesterol", "Cholesterol Level", "Blood Pressure - HeartAttackData", "Blood Pressure - HeartDiseaseData", "Family History", "Family Heart Disease"]]
```

**Right Join**<br>
Show all the patients from the heart disease data (right table) and only those that match from the heart attack risk data (left table).
- Display all the columns
- Display a subset of the columns - Age, Sex, Cholesterol level, Blood Pressure, Heart Rate, Family History

```python
# Show all the patients from the heart disease data (right table) and only those that match from the heart attack risk data (left table) and display all the columns
right_join_on_datasets = heart_att.merge(heart_disease_with_patient_id, how="right", on="Patient ID", suffixes=[" - HeartAttackData", " - HeartDiseaseData"])
right_join_on_datasets #66 rows
```

```python
# Show all the patients from the heart disease data (right table) and only those that match from the heart attack risk data (left table) and display a subset of the columns  
# Age, Gender, Cholesterol level, Blood Pressure, Family History
right_join_on_datasets[["Patient ID", "Age - HeartAttackData", "Gender", "Cholesterol", "Cholesterol Level", "Blood Pressure - HeartAttackData", "Blood Pressure - HeartDiseaseData", "Family History", "Family Heart Disease"]]
```

**Outer Join**<br>
Show all the patients from both heart attack risk and heart disease data and match them where possible
- Display all the columns
- Display a subset of the columns - Age, Sex, Cholesterol level, Blood Pressure, Heart Rate, Family History

```python
# Show all the patients from both heart attack risk and heart disease data and match them where possible and display all the columns
outer_join_on_datasets = heart_att.merge(heart_disease_with_patient_id, how="outer", on="Patient ID", suffixes=[" - HeartAttackData", " - HeartDiseaseData"])
outer_join_on_datasets #8,793 rows
```

```python
# Show all the patients from both heart attack risk and heart disease data and match them where possible and display a subset of the columns:  
# Age, Gender, Cholesterol level, Blood Pressure, Family History
outer_join_on_datasets[["Patient ID", "Age - HeartAttackData", "Gender", "Cholesterol", "Cholesterol Level", "Blood Pressure - HeartAttackData", "Blood Pressure - HeartDiseaseData", "Family History", "Family Heart Disease"]]
```
